```python
def size_evaluater(file_path):
    with open(file_path) as f:
        polygon_data = json.load(f)
    
    coordinates=ast.literal_eval(json.dumps(polygon_data["features"][0]["geometry"]['coordinates']))
    poly=Polygon(coordinates[0])
    
    gdf = gpd.GeoDataFrame({'geometry':[poly]}, crs="EPSG:4326")
    gdf_utm = gdf.to_crs(epsg=32644)  # choose appropriate UTM zone
    
    area_m2 = gdf_utm['geometry'].area[0]
    area_km2=area_m2/1e6
    
    if area_km2*125509 < 12550900:
        prompt="⚠️ AOI too small! Must cover at least ~100 KM^2"
    elif area_km2*125509 > 50331648:
        prompt="⚠️ AOI too large! Must be under ~401 KM^2"
    else:
        prompt="✅ AOI size acceptable."
    os.remove(file_path)
    return coordinates,float(area_m2),float(area_km2),prompt
```

```python
# Initialize Earth Engine

ee.Initialize()

# -----------------------------
# Step 1: Define AOI
# -----------------------------


AOI = ee.Geometry.Polygon(
    [[[
              83.27291159774688,
              17.708606849447705
            ],
            [
              83.33877025100139,
              17.673192393593524
            ],
            [
              83.5331435379859,
              17.90734050151822
            ],
            [
              83.46238244137402,
              17.943863614983712
            ],
            [
              83.27291159774688,
              17.708606849447705
            ]]]
)

# -----------------------------
# Step 2: Function to get yearly NDWI
# -----------------------------
def get_yearly_ndwi(year):
    start = ee.Date.fromYMD(year, 1, 1)
    end = ee.Date.fromYMD(year, 5, 31)

    dataset = "COPERNICUS/S2" if year <= 2018 else "COPERNICUS/S2_SR_HARMONIZED"

    collection = (ee.ImageCollection(dataset)
                  .filterBounds(AOI)
                  
                  .filterDate(start, end)
                  .filter(ee.Filter.lt("CLOUDY_PIXEL_PERCENTAGE", 5)))

    size = collection.size().getInfo()
    if size == 0:
        composite = (ee.Image.constant([0, 0, 0, 0])
                     .rename(["B3", "B4", "B8", "B8A"])  # add required bands
                     .clip(AOI))
    else:
        composite = collection.median()
        # ensure required bands exist
        for b in ["B3", "B4", "B8"]:
            if not composite.bandNames().contains(b).getInfo():
                composite = composite.addBands(ee.Image.constant(0).rename(b))

    # NDWI = (Green - NIR) / (Green + NIR)
    ndwi = composite.normalizedDifference(["B3", "B8"]).rename("NDWI")

    # NDVI = (NIR - Red) / (NIR + Red)
    ndvi = composite.normalizedDifference(["B8", "B4"]).rename("NDVI")

    # Mask out vegetation (NDVI > 0.3 for strong vegetation)
    water_masked = ndwi.updateMask(ndvi.lt(0.1))

    return water_masked.set("year", year)

def smooth_contour(contour, window=5):
    smoothed = contour.copy()
    for i in range(2):  # x and y
        smoothed[:, i] = np.convolve(contour[:, i], np.ones(window)/window, mode='same')
    return smoothed

# -----------------------------
# Step 3: Local folder to save images
# -----------------------------

local_folder = "output"
os.makedirs(local_folder, exist_ok=True)

# -----------------------------
# Step 4: Download NDWI images locally
# -----------------------------

years = list(range(2016, 2026))
print(years)
for year in years:
    ndwi = get_yearly_ndwi(year)
    
    print("Year:",year)

    # Get download URL
    url = ndwi.getDownloadURL({
        'scale': 10,
        'region': AOI,
        'format': 'GeoTIFF'
    })
    
    local_path = os.path.join(local_folder, f"NDWI_{year}.tif")
    
    # Download the file
    r = requests.get(url)
    with open(local_path, 'wb') as f:
        f.write(r.content)
    
    print(f"Downloaded NDWI_{year}.tif to {local_folder}")

# -----------------------------
# Step 5: Compute shoreline and plot
# -----------------------------

ndwi_files = [f for f in os.listdir(local_folder) if f.endswith('.tif')]
    
for file_name in ndwi_files:
    file_path = os.path.join(local_folder, file_name)
    
    try:
        with rasterio.open(file_path) as src:
            ndwi = src.read(1)
    except rasterio.errors.RasterioIOError:
        print(f"⚠ Skipping unreadable file: {file_name}")
        continue  # skip this file and move to the next
    
    ndwi_smooth = median_filter(ndwi, size=1)
    
    
    valid_pixels = ndwi_smooth[~np.isnan(ndwi_smooth)]
    
    
    if len(valid_pixels) > 0:
        th = filters.threshold_otsu(valid_pixels)
    else:
        th = 0  # fallback
    
    water_mask = ndwi_smooth > th
    
    
    contours = measure.find_contours(water_mask, 0.5)
    
   
    plt.figure(figsize=(8,6))
    plt.imshow(ndwi, cmap='Blues', vmin=-1, vmax=1,interpolation="bilinear")
    
    for contour in contours:
        contour_smooth = smooth_contour(contour, window=1)
        plt.plot(contour_smooth[:, 1], contour_smooth[:, 0], color='red', linewidth=1.5)
    
    plt.title(f"Shoreline for {file_name}")
    plt.axis('off')
    plt.show()
```

```python

# -----------------------------
# Parameters
# -----------------------------
SMOOTH_SIZE = 2          # median filter size
MORPH_SIZE = 2            # morphological iterations
MIN_REGION_PIXELS = 1000   # remove small regions
MIN_CONTOUR_LENGTH = 50   # drop short contours
KEEP_LARGEST_WATER = True
BORDER_PIXELS = 1        # narrow border to ignore edge artifacts (in pixels)
OUTPUT_FOLDER = "cleaned_output"  # folder to save cleaned TIFFs
os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# Morphological structuring element (3x3 connectivity)
struct = generate_binary_structure(2, 2)

def clean_water_mask(water_mask, min_pixels=MIN_REGION_PIXELS, morph_size=MORPH_SIZE, keep_largest=False):
    opened = binary_opening(water_mask, structure=struct, iterations=morph_size)
    closed = binary_closing(opened, structure=struct, iterations=morph_size)
    labeled, ncomp = label(closed)
    clean = np.zeros_like(closed, dtype=bool)
    if ncomp == 0:
        return clean
    for region_id in range(1, ncomp+1):
        mask_region = (labeled == region_id)
        if mask_region.sum() >= min_pixels:
            clean |= mask_region
    if keep_largest:
        labeled2, n2 = label(clean)
        if n2 > 0:
            sizes = [(rid, (labeled2 == rid).sum()) for rid in range(1, n2+1)]
            rid_max = max(sizes, key=lambda x: x[1])[0]
            clean = (labeled2 == rid_max)
    return clean

def filter_contours_by_length(contours, min_length=MIN_CONTOUR_LENGTH):
    return [c for c in contours if len(c) >= min_length]

folder_path = "output"
for file_name in os.listdir(folder_path):
    file_path = os.path.join(folder_path, file_name)
    if not file_name.lower().endswith((".tif", ".tiff")):
        continue
    try:
        with rasterio.open(file_path) as src:
            ndwi = src.read(1)
            profile = src.profile
            print(f"✅ Opened: {file_name}")
    except rasterio.errors.RasterioIOError:
        print(f"⚠ Skipping unreadable: {file_name}")
        with rasterio.open(os.path.join(OUTPUT_FOLDER, file_name), "r+") as src:
            data = src.read()
            data[:] = 0  # set all pixels to 0 (or np.nan)
            src.write(data)
        print("✅ Replaced with null .tif")

    # 1) Smooth
    ndwi_smooth = median_filter(ndwi, size=SMOOTH_SIZE)

    # 2) Threshold
    valid = ndwi_smooth[~np.isnan(ndwi_smooth)]
    th = threshold_otsu(valid) if valid.size > 0 else 0
    water_mask = ndwi_smooth > th

    # 3) Remove border artifacts
    water_mask[:BORDER_PIXELS, :] = 0
    water_mask[-BORDER_PIXELS:, :] = 0
    water_mask[:, :BORDER_PIXELS] = 0
    water_mask[:, -BORDER_PIXELS:] = 0

    # 4) Clean mask
    water_mask_clean = clean_water_mask(water_mask)

    # 5) Contours
    contours = measure.find_contours(water_mask_clean, 0.5)
    contours = filter_contours_by_length(contours)

    good_contours = []
    for c in contours:
        coords = [(pt[1], pt[0]) for pt in c]
        try:
            poly = Polygon(coords)
            if not poly.is_valid:
                poly = poly.buffer(0)
            if poly.area > 1.0:
                good_contours.append(c)
        except Exception:
            continue

    # Save cleaned TIFF
    optimized = (water_mask_clean.astype(np.uint8) * 255)
    profile.update(dtype=rasterio.uint8, count=1, compress='lzw')
    with rasterio.open(os.path.join(OUTPUT_FOLDER, file_name), "w", **profile) as dst:
        dst.write(optimized, 1)
    print(f"💾 Saved cleaned: {file_name}")

    
    plt.figure(figsize=(8,6))
    plt.imshow(ndwi, cmap='Blues', vmin=-1, vmax=1)
    for contour in good_contours:
        plt.plot(contour[:,1], contour[:,0], color='red', linewidth=1.5)
    plt.title(f"Shoreline cleaned: {file_name}")
    plt.axis('off')
    plt.savefig(os.path.join(OUTPUT_FOLDER, f"plot_{os.path.splitext(file_name)[0]}.png"),
                bbox_inches='tight', dpi=150)
    plt.show()
```

```python


def auto_crop_black_borders_geo(input_tif, edge_frac_thresh=0.01, buffer_pixels=0):
    """
    Crop uniform black borders (zero or nodata) from a georeferenced TIFF and overwrite it.
    - edge_frac_thresh: minimum fraction of non-black pixels in a row/col to be considered "content"
      (e.g., 0.01 means row/col must have >1% non-black pixels to be kept).
    - buffer_pixels: optional number of pixels to expand the crop (use 0 to be tight).
    """
    folder, filename = os.path.split(input_tif)
    tmp_path = os.path.join(folder, f"tmp_{filename}")

    with rasterio.open(input_tif) as src:
        profile = src.profile.copy()
        bands = src.count

        # Read all bands (shape: bands, rows, cols)
        img = src.read()  # uint8/uint16/float -> numpy array
        rows, cols = img.shape[1], img.shape[2]

        # Determine what is "black"/background:
        # If nodata defined, treat nodata as black; otherwise treat value == 0 as black.
        nodata = src.nodata
        if nodata is None:
            # For multi-band images, treat pixel as non-black if ANY band > 0
            nonblack_mask = np.any(img > 0, axis=0)
        else:
            nonblack_mask = np.any(img != nodata, axis=0)

        # Compute fraction of non-black pixels per row and per column
        row_nonblack_frac = np.mean(nonblack_mask, axis=1)  # length = rows
        col_nonblack_frac = np.mean(nonblack_mask, axis=0)  # length = cols

        # Rows/cols considered content if fraction > edge_frac_thresh
        row_inds = np.where(row_nonblack_frac > edge_frac_thresh)[0]
        col_inds = np.where(col_nonblack_frac > edge_frac_thresh)[0]

        if row_inds.size == 0 or col_inds.size == 0:
            print(f"⚠️ {filename}: no content detected with threshold={edge_frac_thresh}. Skipping.")
            return False

        rmin, rmax = int(row_inds[0]), int(row_inds[-1])
        cmin, cmax = int(col_inds[0]), int(col_inds[-1])

        # apply buffer (expand)
        rmin = max(rmin - buffer_pixels, 0)
        rmax = min(rmax + buffer_pixels, rows - 1)
        cmin = max(cmin - buffer_pixels, 0)
        cmax = min(cmax + buffer_pixels, cols - 1)

        new_height = rmax - rmin + 1
        new_width = cmax - cmin + 1

        # Debug info
        print(f"[DEBUG] {filename}: orig size = ({rows}, {cols})")
        print(f"[DEBUG] content rows: {rmin} -> {rmax}  cols: {cmin} -> {cmax}")
        print(f"[DEBUG] new size = ({new_height}, {new_width})  threshold={edge_frac_thresh} buffer={buffer_pixels}")

        # Read window using rasterio windows (close src automatically after with)
        window = rasterio.windows.Window.from_slices((rmin, rmax + 1), (cmin, cmax + 1))
        out_image = src.read(window=window)
        out_transform = src.window_transform(window)

        # Update profile and write to temp file
        profile.update({
            "height": out_image.shape[1],
            "width": out_image.shape[2],
            "transform": out_transform
        })

    # Write temp file (after input closed)
    with rasterio.open(tmp_path, "w", **profile) as dst:
        dst.write(out_image)

    # Replace original safely (backup option could be added)
    shutil.move(tmp_path, input_tif)
    print(f"✅ Cropped & overwritten: {input_tif}")
    return True


def batch_auto_crop_folder(folder, edge_frac_thresh=0.01, buffer_pixels=0):
    """
    Process all .tif files in folder. Default edge_frac_thresh=0.01 = 1% threshold.
    If cropping too little (borders remain), increase edge_frac_thresh (e.g., 0.02 or 0.05).
    If cropping too aggressively, decrease it (e.g., 0.005).
    """
    for fn in os.listdir(folder):
        if fn.lower().endswith(".tif"):
            path = os.path.join(folder, fn)
            try:
                auto_crop_black_borders_geo(path, edge_frac_thresh=edge_frac_thresh, buffer_pixels=buffer_pixels)
            except Exception as e:
                print(f"❌ Error cropping {fn}: {e}")

# Example usage:
batch_auto_crop_folder("cleaned_output", edge_frac_thresh=0.005, buffer_pixels=0)
```

shorelines extraction with no borders

```python


# -----------------------------
# Parameters
# -----------------------------
folder = "cleaned_output"  # folder containing NDWI GeoTIFFs
tif_files = sorted(glob.glob(os.path.join(folder, "*.tif")))
shorelines = []
last_crs = None  # keep track of raster CRS

# -----------------------------
# NDWI → Shoreline extraction
# -----------------------------
for tif_file in tif_files:
    fname = os.path.basename(tif_file).replace(".tif", "")
    parts = fname.split("_")
    year = int(parts[1])
    month = int(parts[2]) if len(parts) > 2 else None

    try:
        with rasterio.open(tif_file) as src:
            ndwi = src.read(1).astype(float)
            transform = src.transform
            last_crs = src.crs
    except Exception as e:
        print(f"Skipping {tif_file}: {e}")
        continue

    # -----------------------------
    # Step 1: Smooth NDWI
    # -----------------------------
    ndwi_smooth = median_filter(ndwi, size=3)  # kernel size can be adjusted

    # -----------------------------
    # Step 2: Adaptive threshold using Otsu
    # -----------------------------
    valid_pixels = ndwi_smooth[~np.isnan(ndwi_smooth)]
    th = filters.threshold_otsu(valid_pixels) if len(valid_pixels) > 0 else 0
    water_mask = ndwi_smooth > th

    # -----------------------------
    # Step 3: Detect contours
    # -----------------------------
    contours = measure.find_contours(water_mask.astype(float), 0.5)

    for contour in contours:
        # Convert row,col to x,y
        coords = [rasterio.transform.xy(transform, int(r), int(c)) for r, c in contour]
        if len(coords) > 5:
            shorelines.append({
                "year": year,
                "month": month,
                "geometry": LineString(coords)
            })

# -----------------------------
# Step 4: Save GeoJSON
# -----------------------------
print(shorelines)
if shorelines:
    raster_crs = last_crs.to_string() if last_crs else "EPSG:4326"
    gdf = gpd.GeoDataFrame(shorelines, geometry="geometry", crs=raster_crs)

    # Optionally project to UTM (meters)
    gdf = gdf.to_crs(epsg=32644)

    gdf.to_file("shorelines.geojson", driver="GeoJSON")
    print(f"✅ Extracted {len(gdf)} shorelines saved to shorelines.geojson")
else:
    print("⚠ No shorelines found in the NDWI images.")
import matplotlib.pyplot as plt

# Ensure shorelines were extracted
if len(gdf) == 0:
    print("⚠ No shorelines to plot")
else:
    # Suppose your GeoDataFrame has a 'year' column
    years = sorted(gdf['year'].unique())
    
    # Colormap (distinct colors for each year)
    cmap = cm.get_cmap("tab10", len(years))
    
    fig, ax = plt.subplots(figsize=(12, 8))
    
    for i, year in enumerate(years):
        gdf_year = gdf[gdf['year'] == year]
        gdf_year.plot(ax=ax, color=cmap(i), linewidth=0.5, alpha=0.5,label=str(year))
    
    # Fancy styling
    ax.set_title("Shoreline Change (2016–2025)", fontsize=18, fontweight="bold", pad=15)
    ax.legend(title="Year", fontsize=10, title_fontsize=12, loc="center left", bbox_to_anchor=(1, 0.5))
    ax.set_xlabel("Easting (m)", fontsize=12)
    ax.set_ylabel("Northing (m)", fontsize=12)
    ax.grid(True, linestyle="--", alpha=0.5)  # nice grid
    plt.tight_layout()
    plt.show()
```

```python
def split_shorelines_by_year(input_file, output_folder):
    """
    Splits a shoreline GeoJSON by year and saves separate files.
    """
    # Load shoreline dataset
    gdf = gpd.read_file(input_file)

    # Make sure folder exists
    os.makedirs(output_folder, exist_ok=True)

    # Loop through years automatically
    for year in sorted(gdf['year'].unique()):
        yearly = gdf[gdf['year'] == year]
        out_file = os.path.join(output_folder, f"shoreline_{year}.geojson")
        yearly.to_file(out_file, driver="GeoJSON")
        print(f"✅ Saved shoreline for {year} -> {out_file}")

# ------------------------
# Example usage
# ------------------------
input_file = "shorelines.geojson"   # file with all years
output_folder = "shorelines_data"

split_shorelines_by_year(input_file, output_folder)
```

Transect generation with no Borders!

```python


# -----------------------------
# 1️⃣ Load Shorelines
# -----------------------------
l
gdf = gpd.read_file("shorelines.geojson")
gdf = gdf.to_crs(epsg=32644)  # Project to UTM for meters
print(f"✅ Loaded {len(gdf)} shoreline geometries")

# -----------------------------
# 2️⃣ Generate Baseline
# -----------------------------
early_years = gdf[gdf['year'].isin(sorted(gdf['year'].unique())[:3])]
baseline_geom = early_years.unary_union
baseline_gdf = gpd.GeoDataFrame(geometry=[baseline_geom], crs=gdf.crs)
baseline_gdf.to_file("baseline_shoreline.geojson", driver="GeoJSON")
print("✅ Baseline shoreline saved to 'baseline_shoreline.geojson'")

# -----------------------------
# 3️⃣ Generate Transects
# -----------------------------
transects = []
transect_length = 2000  # meters
spacing = 100  # meters between transects

lines = [baseline_geom] if isinstance(baseline_geom, LineString) else list(baseline_geom.geoms)
for line in lines:
    line_len = line.length
    n_transects = max(int(line_len // spacing), 1)
    for i in range(n_transects):
        d = i * spacing
        step = min(spacing / 10, line_len - d)
        if step <= 0:
            continue
        p = line.interpolate(d)
        p_next = line.interpolate(d + step)
        dx, dy = p_next.x - p.x, p_next.y - p.y
        length = math.hypot(dx, dy)
        if length == 0:
            continue
        nx, ny = -dy / length, dx / length
        x1, y1 = p.x - nx * transect_length / 2, p.y - ny * transect_length / 2
        x2, y2 = p.x + nx * transect_length / 2, p.y + ny * transect_length / 2
        transects.append(LineString([(x1, y1), (x2, y2)]))

transects_gdf = gpd.GeoDataFrame({"Transect": range(len(transects)), "geometry": transects}, crs=gdf.crs)
print(f"✅ Generated {len(transects_gdf)} transects")

# -----------------------------
# 4️⃣ Remove Transects Along Borders
# -----------------------------
# AOI bounds (replace with your AOI polygon if you have one)
print(f"✅ {len(transects_gdf)} transects")
transects_gdf.to_file("transects.geojson", driver="GeoJSON")

# -----------------------------
# 5️⃣ Helper: Extract intersection points
# -----------------------------
def extract_points(inter):
    pts = []
    if inter.is_empty:
        return pts
    if inter.geom_type == "Point":
        pts.append(inter)
    elif inter.geom_type == "MultiPoint":
        pts.extend(list(inter.geoms))
    elif inter.geom_type == "LineString":
        pts.append(Point(inter.coords[0]))
        pts.append(Point(inter.coords[-1]))
    elif inter.geom_type == "MultiLineString":
        for line in inter.geoms:
            pts.append(Point(line.coords[0]))
            pts.append(Point(line.coords[-1]))
    elif inter.geom_type == "GeometryCollection":
        for geom in inter.geoms:
            pts.extend(extract_points(geom))
    return pts

# -----------------------------
# 6️⃣ Extract Shoreline Distances
# -----------------------------
gdf_sindex = gdf.sindex
records = []

for t_idx, transect in enumerate(transects_gdf.geometry):
    candidates_idx = list(gdf_sindex.intersection(transect.bounds))
    candidates = gdf.iloc[candidates_idx]
    for year, group in candidates.groupby("year"):
        distances = []
        for shoreline in group.geometry:
            inter = shoreline.intersection(transect)
            pts = extract_points(inter)
            distances.extend([transect.project(pt) for pt in pts])
        if distances:
            records.append({
                "Year": year,
                "Transect": t_idx,
                "Shoreline_Distance": max(distances)
            })

df = pd.DataFrame(records)
df.to_csv("shoreline_timeseries_fast.csv", index=False)
print(f"✅ Extracted {len(df)} records and saved to shoreline_timeseries_fast.csv")

# -----------------------------
# 7️⃣ Optional: Quick Plot
# -----------------------------
fig, ax = plt.subplots(figsize=(12, 8))
ax.set_facecolor("#f0f8ff")
gdf[gdf['year']==gdf['year'].max()].plot(ax=ax, color='royalblue', linewidth=1, label='Latest Shoreline')
baseline_gdf.plot(ax=ax, color="crimson", linewidth=1, linestyle="--", label="Baseline")
transects_gdf.plot(ax=ax, color='black', linewidth=0.2, alpha=0.05, label='Transects')
plt.title("Shoreline – Baseline – Transects", fontsize=16, fontweight="bold", pad=15)
plt.xlabel("Longitude", fontsize=12)
plt.ylabel("Latitude", fontsize=12)
ax.grid(color="gray", linestyle=":", linewidth=0.5, alpha=0.7)
plt.legend(frameon=True, facecolor="white", edgecolor="black", fontsize=10, loc="lower right")
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
plt.show()
```

Comparing Validation with previous years data with different degrees and choosing the right degree
#### HYPER PARAMETER TUNING

```python



# -----------------------------
# 1. Load shorelines & transects
# -----------------------------
shorelines = gpd.read_file("shorelines.geojson").to_crs(epsg=32644)
transects = gpd.read_file("transects.geojson").to_crs(epsg=32644)

print(f"✅ Loaded {len(shorelines)} shorelines, {len(transects)} transects")

# -----------------------------
# 2. Load time-series distances
# -----------------------------
df = pd.read_csv("shoreline_timeseries_fast.csv")
df_yearly = df.groupby(["Transect", "Year"])["Shoreline_Distance"].mean().reset_index()

# -----------------------------
# 3. Validation setup
# -----------------------------
validation_years = [2017,  2020, 2025]   # 🔹 years to validate
degrees = [1, 2, 3, 4]                     # 🔹 polynomial degrees to test

results = []
best_per_year = {}   # store best model info per year

for val_year in validation_years:
    best_mae, best_rmse, best_degree = float("inf"), None, None

    for degree in degrees:
        y_true, y_pred = [], []

        for t_idx in df_yearly["Transect"].unique():
            df_t = df_yearly[df_yearly["Transect"] == t_idx]
            X = df_t["Year"].values.reshape(-1, 1)
            y = df_t["Shoreline_Distance"].values

            if len(y) < degree + 1:  # need enough points for fitting
                continue

            poly = PolynomialFeatures(degree=degree)
            X_poly = poly.fit_transform(X)

            model = LinearRegression()
            model.fit(X_poly, y)

            if val_year in df_t["Year"].values:
                true_dist = df_t.loc[df_t["Year"] == val_year, "Shoreline_Distance"].values[0]
                pred_dist = model.predict(poly.transform([[val_year]]))[0]

                y_true.append(true_dist)
                y_pred.append(pred_dist)

        if len(y_true) > 0:
            mae = mean_absolute_error(y_true, y_pred)
            rmse = sqrt(mean_squared_error(y_true, y_pred))
            results.append({"Degree": degree, "Validation_Year": val_year, "MAE": mae, "RMSE": rmse})
            print(f"✅ Validation {val_year} (Degree={degree}): MAE={mae:.2f} m, RMSE={rmse:.2f} m")

            # 🔹 Auto-degree tuning (choose lowest MAE)
            if mae < best_mae:
                best_mae, best_rmse, best_degree = mae, rmse, degree

    best_per_year[val_year] = {"Best_Degree": best_degree, "MAE": best_mae, "RMSE": best_rmse}
    print(f"🎯 Best Degree for {val_year}: {best_degree} (MAE={best_mae:.2f}, RMSE={best_rmse:.2f})\n")

# -----------------------------
# 4. Show results as table
# -----------------------------
results_df = pd.DataFrame(results)
print("\n📊 Comparison Table (MAE):")
print(results_df.pivot(index="Validation_Year", columns="Degree", values="MAE"))
print("\n📊 RMSE Comparison:")
print(results_df.pivot(index="Validation_Year", columns="Degree", values="RMSE"))

print("\n🏆 Auto-Tuned Best Degrees per Year:")
print(pd.DataFrame(best_per_year).T)

# -----------------------------
# 5. Auto-select best degree
# -----------------------------

# Compute average MAE across validation years for each degree
avg_mae = results_df.groupby("Degree")["MAE"].mean()
avg_rmse = results_df.groupby("Degree")["RMSE"].mean()

best_degree_mae = avg_mae.idxmin()
best_degree_rmse = avg_rmse.idxmin()

print("\n🏆 Auto-Tuning Results:")
print(f"Best Degree (by MAE): {best_degree_mae} with avg MAE = {avg_mae.min():.2f} m")
print(f"Best Degree (by RMSE): {best_degree_rmse} with avg RMSE = {avg_rmse.min():.2f} m")
```

VALIDATION using MAE and RMSE

```python
# -----------------------------
# 1. Load data
# -----------------------------

shorelines = gpd.read_file("shorelines.geojson").to_crs(epsg=32644)
transects = gpd.read_file("transects.geojson").to_crs(epsg=32644)
df = pd.read_csv("shoreline_timeseries_fast.csv")

# Compute yearly mean distance per transect
df_yearly = df.groupby(["Transect", "Year"])["Shoreline_Distance"].mean().reset_index()

# -----------------------------
# 2. Pick a past year to validate
# -----------------------------
validation_year = 2025

pred_points = []
actual_points = []

for t_idx in df_yearly["Transect"].unique():
    df_t = df_yearly[df_yearly["Transect"] == t_idx]
    
    # Separate training (past years) and actual (validation year)
    df_train = df_t[df_t["Year"] < validation_year]
    df_actual = df_t[df_t["Year"] == validation_year]
    
    if len(df_train) < 2 or df_actual.empty:
        continue
    
    # -----------------------------
    # 3. Train 4th-degree polynomial model
    # -----------------------------
    X_train = df_train["Year"].values
    y_train = df_train["Shoreline_Distance"].values
    coeffs = np.polyfit(X_train, y_train, deg=poly_degree)   # degree 4
    poly_model = np.poly1d(coeffs)
    
    # -----------------------------
    # 4. Predict for validation year
    # -----------------------------
    dist_pred = poly_model(validation_year)
    row = transects.loc[transects["Transect"] == t_idx]
    if row.empty:
        count+=1
        continue
        
    tran = row.iloc[0].geometry    
    point_pred = tran.interpolate(dist_pred)
    
    # Actual point
    dist_actual = df_actual["Shoreline_Distance"].values[0]
    point_actual = tran.interpolate(dist_actual)
    
    pred_points.append(point_pred)
    actual_points.append(point_actual)
print(count)
# -----------------------------
# 5. Compute error metrics
# -----------------------------
pred_coords = np.array([[p.x, p.y] for p in pred_points])
actual_coords = np.array([[p.x, p.y] for p in actual_points])

# Euclidean distance per transect
errors = np.linalg.norm(pred_coords - actual_coords, axis=1)
mae = np.mean(errors)
rmse = np.sqrt(np.mean(errors**2))

print(f"✅ Validation for {validation_year}: MAE = {mae:.2f} m, RMSE = {rmse:.2f} m")

# -----------------------------
# 6. Convert to GeoDataFrames for plotting
# -----------------------------
pred_gdf = gpd.GeoDataFrame(geometry=pred_points, crs=transects.crs)
actual_gdf = gpd.GeoDataFrame(geometry=actual_points, crs=transects.crs)

# -----------------------------
# 7. Plot
# -----------------------------
fig, ax = plt.subplots(figsize=(12, 10))
shorelines[shorelines["year"] == validation_year].plot(ax=ax, color="blue", markersize=0.5, label="Actual Shoreline")
transects.plot(ax=ax, color="gray", linewidth=0.2, alpha=0.2)
pred_gdf.plot(ax=ax, color="red", markersize=3, label="Predicted Points")
actual_gdf.plot(ax=ax, color="green", markersize=3, label="Actual Points")

plt.legend()
plt.title(f"Shoreline Prediction Validation for {validation_year}")
plt.xlabel("Easting (m)")
plt.ylabel("Northing (m)")
plt.show()
```

EROSION RATE AND ACCRETION RATE in (m/yr)

```python


# -----------------------------
# Step 1: Load Shorelines + Transects
# -----------------------------
shoreline_base = gpd.read_file(os.path.join("shorelines_data","shoreline_2016.geojson")).to_crs(epsg=32644)
shoreline_future = gpd.read_file(os.path.join("shorelines_data","shoreline_2025.geojson")).to_crs(epsg=32644)
transects = gpd.read_file("transects.geojson").to_crs(epsg=32644)

# Merge shoreline geometries into a single line
shoreline_base_union = shoreline_base.geometry.union_all()   # Updated from unary_union
shoreline_future_union = shoreline_future.geometry.union_all()  # Updated from unary_union

# -----------------------------
# Step 2: Helper to safely get a Point from intersection
# -----------------------------
def get_point(inter, ref_line):
    """Return a single Point from an intersection, handling MultiPoint or GeometryCollection"""
    if inter.is_empty:
        return None
    elif inter.geom_type == "Point":
        return inter
    elif inter.geom_type in ["MultiPoint", "GeometryCollection"]:
        # take the point closest to transect centroid
        pts = []
        if inter.geom_type == "MultiPoint":
            pts = list(inter.geoms)
        elif inter.geom_type == "GeometryCollection":
            for g in inter.geoms:
                if g.geom_type == "Point":
                    pts.append(g)
        if pts:
            return min(pts, key=lambda p: p.distance(ref_line.centroid))
    elif inter.geom_type == "LineString":
        # take midpoint of line
        return Point(inter.coords[int(len(inter.coords)/2)])
    return None

# -----------------------------
# Step 3: Calculate change along transects
# -----------------------------
results = []

for idx, tr in transects.iterrows():
    base_pt = get_point(tr.geometry.intersection(shoreline_base_union), tr.geometry)
    fut_pt = get_point(tr.geometry.intersection(shoreline_future_union), tr.geometry)

    if base_pt and fut_pt:
        dist = base_pt.distance(fut_pt)
        # sign based on relative y-direction
        sign = 1 if fut_pt.y > base_pt.y else -1
        dist_signed = sign * dist

        results.append({
            "transect_id": idx,
            "base_year": 2016,
            "future_year": 2025,
            "erosion_accretion_m": dist_signed
        })

# Calculate rate per year
for r in results:
    years_diff = r["future_year"] - r["base_year"]
    r["rate_m_per_year"] = r["erosion_accretion_m"] / years_diff if years_diff > 0 else None

print("Rate --> (m/year) is calculated")

# -----------------------------
# Step 4: Save results
# -----------------------------
df = pd.DataFrame(results)
df.to_csv("erosion_accretion_results.csv", index=False)
print("✅ Erosion/Accretion analysis completed. Saved to erosion_accretion_results.csv")
print(df)
```

```python
mean_rate = df["rate_m_per_year"].mean()
median_rate = df["rate_m_per_year"].median()
erosion_percent = (df["rate_m_per_year"] < 0).mean() * 100
accretion_percent = (df["rate_m_per_year"] > 0).mean() * 100

print("Mean rate:", mean_rate, "m/year")
print("Median rate:", median_rate, "m/year")
print(f"Eroding transects: {erosion_percent}%")
print(f"Accreting transects: {accretion_percent}%")
print(f"stable transects: {100-(accretion_percent+erosion_percent)}%")
```
